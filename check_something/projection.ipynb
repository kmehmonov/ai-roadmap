```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # 3D plotting

# Create grid
x = np.linspace(-2, 2, 100)
y = np.linspace(-2, 2, 100)
X, Y = np.meshgrid(x, y)

# Define the exponential surface: z = e^(-x^2 - y^2)
Z = np.exp(-(X**2 + Y**2))

# Create a 3D figure
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot the surface
surf = ax.plot_surface(X, Y, Z, cmap='viridis', alpha=0.8, edgecolor='none')

# Add contours projected to the XY plane (zdir='z' means project along z-axis)
contours = ax.contour(X, Y, Z, zdir='z', offset=1, cmap='viridis', levels = 20)

# Labels and styling
ax.set_title('Exponential Surface with XY Projection')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_zlim(0, 1.1)

plt.tight_layout()
plt.show()
```
